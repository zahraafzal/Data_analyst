# Sales Data Analysis Project
**Data Analyst Project — Retail Sales Dataset**

In this project, we will analyze sales data from a retail store. The objective is to clean the data and extract useful insights — such as which category generates the most sales, which region performs best, and how sales trends change over time.

**Tools used:** Python, Pandas, Matplotlib


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Display settings
pd.set_option('display.max_columns', None)
plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. Loading the Dataset

```python
df = pd.read_csv("sales_data.csv")
df.head()
```

```python
df.info()
```

## 2. Data Cleaning

First, we will check if there are any missing values or duplicate rows in the data.

```python
# Check missing values
df.isnull().sum()
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Remove duplicates
df = df.drop_duplicates()

# Fill missing UnitPrice with the average price of that Product
df['UnitPrice'] = df.groupby('Product')['UnitPrice'].transform(lambda x: x.fillna(x.mean()))

# Fill missing Region with 'Unknown'
df['Region'] = df['Region'].fillna('Unknown')

# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

print("Missing values after cleaning:")
print(df.isnull().sum())
print("\nTotal rows after cleaning:", len(df))
```

## 3. Feature Engineering

We create a new column `TotalSales` by multiplying Quantity and UnitPrice to get the total sales amount. Similarly, we extract a `Month` column for time-based analysis.

```python
df['TotalSales'] = df['Quantity'] * df['UnitPrice']
df['Month'] = df['Date'].dt.to_period('M').astype(str)
df.head()
```

## 4. Basic Statistics

```python
df[['Quantity', 'UnitPrice', 'TotalSales']].describe()
```

## 5. Exploratory Data Analysis (EDA)

### 5.1 Total Sales by Category

```python
category_sales = df.groupby('Category')['TotalSales'].sum().sort_values(ascending=False)
print(category_sales)

plt.figure(figsize=(8,5))
category_sales.plot(kind='bar', color='skyblue')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales (PKR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('sales_by_category.png')
plt.show()
```

### 5.2 Total Sales by Region

```python
region_sales = df.groupby('Region')['TotalSales'].sum().sort_values(ascending=False)
print(region_sales)

plt.figure(figsize=(8,5))
region_sales.plot(kind='bar', color='lightcoral')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales (PKR)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('sales_by_region.png')
plt.show()
```

### 5.3 Sales Trend Over Time

```python
monthly_sales = df.groupby('Month')['TotalSales'].sum()
print(monthly_sales)

plt.figure(figsize=(10,5))
monthly_sales.plot(kind='line', marker='o', color='purple', linewidth=2)
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales (PKR)')
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('sales_trend.png')
plt.show()
```

### 5.4 Top 5 Products by Sales

```python
top_products = df.groupby('Product')['TotalSales'].sum().sort_values(ascending=False).head(5)
print(top_products)

plt.figure(figsize=(8,5))
top_products.plot(kind='barh', color='seagreen')
plt.title('Top 5 Best-Selling Products')
plt.xlabel('Total Sales (PKR)')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.savefig('top_products.png')
plt.show()
```

## 6. Conclusion / Insights

Based on this analysis, we can draw the following conclusions:

- The **Electronics** category generates the highest revenue due to its higher unit prices.
- **Faisalabad** region has the highest sales performance, followed by Karachi.
- Sales trends show variation across months, indicating seasonal demand patterns.
- The top 5 products are the most profitable items for the company and should be prioritized in inventory management.

This simple analysis can help businesses make informed decisions — such as which categories to focus on, which regions need increased marketing efforts, and which products to keep in higher stock.
